#### Question 1

```python
# Given Initial Population and their Decimal Values
initial_population = {
    "0011": 3,
    "0101": 5,
    "1001": 9,
    "1110": 14
}

# 1. Calculate the fitness value f(x) = x^2 for each chromosome
chromosome_fitness = []
print("--- Chromosome Fitness Calculation ---")
for chromosome, x_value in initial_population.items():
    fitness_value = x_value**2
    chromosome_fitness.append({
        "chromosome": chromosome,
        "x_value": x_value,
        "fitness": fitness_value
    })
    print(f"Chromosome: {chromosome}, Decimal Value (x): {x_value}, Fitness (f(x)=x^2): {fitness_value}")

# 2. Rank the chromosomes according to fitness value
# Sort in descending order of fitness
ranked_chromosomes = sorted(chromosome_fitness, key=lambda item: item['fitness'], reverse=True)

print("\n--- Ranked Chromosomes by Fitness ---")
for i, item in enumerate(ranked_chromosomes):
    print(f"Rank {i+1}: Chromosome: {item['chromosome']}, x: {item['x_value']}, Fitness: {item['fitness']}")

# 3. Apply selection and choose the two best chromosomes as parents for the next generation
parent1 = ranked_chromosomes[0]
parent2 = ranked_chromosomes[1]

print("\n--- Selected Parents for Next Generation ---")
print(f"Parent 1 (Best): Chromosome: {parent1['chromosome']}, x: {parent1['x_value']}, Fitness: {parent1['fitness']}")
print(f"Parent 2 (Second Best): Chromosome: {parent2['chromosome']}, x: {parent2['x_value']}, Fitness: {parent2['fitness']}")
```

#### Question 2

```python
# Define the distance table between locations
distance_matrix = {
    'A': {'B': 10, 'C': 15, 'D': 20},
    'B': {'A': 10, 'C': 35, 'D': 25},
    'C': {'A': 15, 'B': 35, 'D': 30},
    'D': {'A': 20, 'B': 25, 'C': 30}
}

def get_route_distance(route):
    total_distance = 0
    for i in range(len(route) - 1):
        total_distance += distance_matrix[route[i]][route[i+1]]
    return total_distance

# Given Initial Population (Routes)
initial_routes = [
    ['A', 'B', 'C', 'D'],
    ['A', 'C', 'B', 'D'],
    ['B', 'A', 'D', 'C'],
    ['C', 'B', 'A', 'D']
]

# 1. Calculate the total distance (fitness) for each route
routes_with_fitness = []
print("--- Route Distance Calculation (Fitness) ---")
for route in initial_routes:
    distance = get_route_distance(route)
    routes_with_fitness.append({
        "route": route,
        "distance": distance
    })
    print(f"Route: {' -> '.join(route)}, Distance: {distance}")

# 2. Rank the chromosomes according to fitness value (shortest distance first)
ranks = sorted(routes_with_fitness, key=lambda item: item['distance'])

print("\n--- Ranked Routes by Distance ---")
for i, item in enumerate(ranks):
    print(f"Rank {i+1}: Route: {' -> '.join(item['route'])}, Distance: {item['distance']}")

# 3. Apply selection and choose the two best routes as parents for the next generation
parent1 = ranks[0]['route']
parent2 = ranks[1]['route']

print("\n--- Selected Parents for Crossover ---")
print(f"Parent 1 (Best): {' -> '.join(parent1)} (Distance: {ranks[0]['distance']})")
print(f"Parent 2 (Second Best): {' -> '.join(parent2)} (Distance: {ranks[1]['distance']})")

# Apply single-point crossover to generate two new offspring routes
crossover_point = 2 # Randomly select kia hai

offspring1 = parent1[:crossover_point]
for city in parent2:
    if city not in offspring1:
        offspring1.append(city)

offspring2 = parent2[:crossover_point]
for city in parent1:
    if city not in offspring2:
        offspring2.append(city)

print("\n--- Offspring Routes after Single-Point Crossover ---")
print(f"Offspring 1: {' -> '.join(offspring1)}")
print(f"Offspring 2: {' -> '.join(offspring2)}")

# 4. Show the new generation of routes.
# Calculate distances for the new offspring
offspring1_distance = get_route_distance(offspring1)
offspring2_distance = get_route_distance(offspring2)

print("\n--- New Generation of Routes (including Offspring and their distances) ---")
print(f"Offspring 1: {' -> '.join(offspring1)}, Distance: {offspring1_distance}")
print(f"Offspring 2: {' -> '.join(offspring2)}, Distance: {offspring2_distance}")
```

```python
import random

# Given Initial Population (arrangements of products)
initial_population_q3 = [
    ['A', 'B', 'C', 'D'],
    ['B', 'C', 'A', 'D'],
    ['D', 'A', 'B', 'C'],
    ['C', 'B', 'D', 'A']
]

print("Initial Population (Product Arrangements):")
for i, arrangement in enumerate(initial_population_q3):
    print(f"Chromosome {i+1}: {' '.join(arrangement)}")

# Mutation
mutated_population = []

for chromosome in initial_population_q3:
    # chromosome copy krrhy
    mutated_chromosome = list(chromosome)

    # Randomly select two distinct positions to swap
    idx1, idx2 = random.sample(range(len(mutated_chromosome)), 2)

    # Perform the swap
    mutated_chromosome[idx1], mutated_chromosome[idx2] = mutated_chromosome[idx2], mutated_chromosome[idx1]

    mutated_population.append(mutated_chromosome)

# new mutated population
print("\nMutated Population:")
for i, arrangement in enumerate(mutated_population):
    print(f"Mutated Chromosome {i+1}: {' '.join(arrangement)}")

# Explanation
print("\n--- Explanation of Mutation's Role ---")
print("Mutation plays a crucial role in Genetic Algorithms, including optimizing product placement, by:")
print("1. Increasing Diversity: It introduces new genetic material (new arrangements) into the population that might not be present through crossover alone. This helps prevent the algorithm from getting stuck in local optima (suboptimal solutions).")
print("2. Escaping Local Optima: By randomly altering product positions, mutation allows the search to explore new areas of the solution space. For product placement, a small change (like swapping two products) might lead to a configuration that, while initially not much better, opens up a path to a significantly superior arrangement.")
print("3. Preventing Premature Convergence: Without mutation, the population might quickly converge to a limited set of solutions. Mutation ensures a continuous exploration, maintaining genetic variability which is essential for finding truly optimal or near-optimal product placements.")
print("In essence, mutation provides the 'innovation' in the genetic algorithm, helping it to find novel and potentially better ways to arrange products for optimal warehouse efficiency.")
```

```python
import random

print("--- Question 4: Problem Solving using Genetic Algorithm ---")

# Define the target expression and evaluation function
def evaluateExpression(x, y, z):
    return 6 * x**3 + 9 * y**2 + 90 * z

# Define the fitness function
def fitness(x, y, z):
    ans = evaluateExpression(x, y, z) - 25
    if ans == 0:
        return 999999999  # Very high fitness for exact solution
    else:
        return 1 / abs(ans)

# Given Initial Population
initial_population_q4 = [
    (1, 1, 0),
    (2, 1, 0),
    (1, 2, 0),
    (2, 2, 0)
]

populationSize = len(initial_population_q4)
solutions_q4 = list(initial_population_q4) # Start with the given population

# Parameters
generations = 5
mutationStrength = 0.1 # +/- 0.1 for mutation

print("Initial Population:")
for sol in solutions_q4:
    print(f"  x={sol[0]}, y={sol[1]}, z={sol[2]}")

for generation_count in range(generations):
    print(f"\n--- Generation {generation_count + 1} ---")

    # 1. Evaluate the expression and 2. Calculate the fitness value for each solution
    ranked_solutions_q4 = []
    print("  Expression Evaluation and Fitness Calculation:")
    for solution in solutions_q4:
        eval_result = evaluateExpression(solution[0], solution[1], solution[2])
        fitness_val = fitness(solution[0], solution[1], solution[2])
        ranked_solutions_q4.append((fitness_val, solution))
        print(f"    Solution ({solution[0]:.2f}, {solution[1]:.2f}, {solution[2]:.2f}): Evaluation = {eval_result:.4f}, Fitness = {fitness_val:.4f}")

    # Sort by fitness (descending)
    ranked_solutions_q4.sort(key=lambda x: x[0], reverse=True)

    # 3. Select the two best solutions based on highest fitness
    parent1_q4 = ranked_solutions_q4[0][1]
    parent2_q4 = ranked_solutions_q4[1][1]

    print("  Selected Best Solutions (Parents):")
    print(f"    Parent 1: ({parent1_q4[0]:.2f}, {parent1_q4[1]:.2f}, {parent1_q4[2]:.2f}), Fitness = {ranked_solutions_q4[0][0]:.4f}")
    print(f"    Parent 2: ({parent2_q4[0]:.2f}, {parent2_q4[1]:.2f}, {parent2_q4[2]:.2f}), Fitness = {ranked_solutions_q4[1][0]:.4f}")

    new_solutions_q4 = []
    # Carry over the best two solutions to the next generation (elitism)
    new_solutions_q4.append(parent1_q4)
    new_solutions_q4.append(parent2_q4)

    # 4. Apply mutation by slightly changing the values of x, y, or z (±0.1) to generate a new population.
    # Fill the rest of the population with mutated versions of the parents
    while len(new_solutions_q4) < populationSize:
        # Choose a random parent for mutation (can be parent1 or parent2)
        parent = random.choice([parent1_q4, parent2_q4])
        mutated_child = list(parent)

        # Mutate each gene (x, y, z) with a slight change of +/- 0.1
        for i in range(3):
            mutated_child[i] += random.choice([-mutationStrength, mutationStrength])
            # Ensure values don't go too low
            mutated_child[i] = max(0, mutated_child[i])

        new_solutions_q4.append(tuple(mutated_child))

    solutions_q4 = new_solutions_q4

print("\n--- Final Best Solution after all Generations ---")
final_best_fitness, final_best_solution_q4 = ranked_solutions_q4[0]
print(f"Best Solution: x={final_best_solution_q4[0]:.4f}, y={final_best_solution_q4[1]:.4f}, z={final_best_solution_q4[2]:.4f}")
print(f"Final Expression Evaluation (6x^3 + 9y^2 + 90z) = {evaluateExpression(final_best_solution_q4[0], final_best_solution_q4[1], final_best_solution_q4[2]):.4f}")
print(f"Corresponding Fitness: {final_best_fitness:.4f}")
```
